## Define the limited model

[Declared] A Boolean pass is `True`; failure is `False`. The defect is
`global verdict - local conjunction`. This helper deliberately rejects integers
and strings so the teaching interface cannot mistake `"False"` for a pass.

This is an illustrative Python function, not a substitute for Rust's typed
`CompositionRecord` constructor, JSON decoding or deployment admission.


```python
def teaching_defect(left_pass, right_pass, global_pass):
    verdicts = (left_pass, right_pass, global_pass)
    if any(type(value) is not bool for value in verdicts):
        raise TypeError("The teaching model accepts exactly three booleans.")
    local_product = left_pass and right_pass
    return int(global_pass) - int(local_product)


def handoff_verdicts(sent, expected):
    if type(sent) is not str or type(expected) is not str:
        raise TypeError("Identifiers in this teaching example are strings.")
    return bool(sent), bool(expected), sent == expected


print("Teaching helpers defined; no native code has run.")
```

```python
sent, expected = "alpha", "beta"
worked_verdicts = handoff_verdicts(sent, expected)
worked_defect = teaching_defect(*worked_verdicts)

assert worked_verdicts == (True, True, False)
assert worked_defect == -1
assert sent != expected
print("Earlier prediction:", prediction_before_evaluation)
print("Supplied identifiers:", sent, expected)
print("Local/global verdicts:", worked_verdicts)
print("Defect:", worked_defect)
print("Both local checks pass, but the equality check fails.")
```

## Enumerate the abstract Boolean carrier

This table contains all possible verdict triples. They need not all be reachable
under the identifier example's particular predicates. The expected rows below
are a handwritten arithmetic reference, separate from the helper's output.
Compare the all-fail and all-pass rows: both have zero defect.


```python
handwritten_reference = {
    (False, False, False): 0,
    (False, False, True): 1,
    (False, True, False): 0,
    (False, True, True): 1,
    (True, False, False): 0,
    (True, False, True): 1,
    (True, True, False): -1,
    (True, True, True): 0,
}
observed = {}
for verdicts in product((False, True), repeat=3):
    observed[verdicts] = teaching_defect(*verdicts)
    assert observed[verdicts] == handwritten_reference[verdicts]
    print(tuple(int(value) for value in verdicts), observed[verdicts])

assert sum(delta == -1 for delta in observed.values()) == 1
assert sum(delta == 0 for delta in observed.values()) == 4
assert sum(delta == 1 for delta in observed.values()) == 3
assert observed[(False, False, False)] == observed[(True, True, True)] == 0
```

## Try plausible wrong rules and malformed inputs

## Feedback after the transfer attempt

Run the next cell only after recording your predictions. Global equality is
deliberately independent of the two local nonempty checks. If the application

```python
transfer_cases = (
    ("alpha", "alpha", (True, True, True), 0),
    ("", "beta", (False, True, False), 0),
    ("", "", (False, False, True), 1),
)
for sent, expected, expected_verdicts, expected_delta in transfer_cases:
    actual_verdicts = handoff_verdicts(sent, expected)
    actual_delta = teaching_defect(*actual_verdicts)
    assert actual_verdicts == expected_verdicts
    assert actual_delta == expected_delta
    print(repr(sent), repr(expected), actual_verdicts, actual_delta)

print("Explain the both-empty +1 using the declared equality predicate.")
```
