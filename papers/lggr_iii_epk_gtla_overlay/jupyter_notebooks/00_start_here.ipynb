### 1. Orientation

When modular systems are assembled from independently verified components, each component can pass its local verification tests while the integrated composition fails its end-to-end specification. In this series, this failure mode is termed **Local Green, Global Red (LGGR)**.

This package models that gap using binary local verdicts $e_1, e_2 \in \{0, 1\}$, a composite verdict $e_g \in \{0, 1\}$, and the signed evaluator gluing defect:

$$\Delta_E = e_g - e_1 e_2 \in \{-1, 0, 1\}.$$

The value $\Delta_E = -1$ is precisely the Local Green, Global Red witness: the local product passes ($e_1 e_2 = 1$), yet the global test fails ($e_g = 0$).

### 2. Glossary of Core Concepts

- **Local Verdicts ($e_1, e_2$):** Binary outcomes of component-level tests ($1 = \text{Pass}, 0 = \text{Fail}$).
- **Local Product ($e_1 e_2$):** Conjunction of local verdicts; passes ($1$) if and only if all local checks pass.
- **Composite Verdict ($e_g$):** Empirical verdict on the end-to-end composed system.
- **Evaluator Defect ($\Delta_E$):** Signed difference $\Delta_E = e_g - e_1 e_2$.
- **Agreement vs Success:** $\Delta_E = 0$ signifies agreement with the local product ($e_g = e_1 e_2$), which includes total failure $(0, 0, 0)$. It does not imply that any component succeeded.
- **The Six Roles:** Orthogonal perspectives interpreting the same observed defect record: Finite Obstruction, Foundations, Representation, Observer, Continuum Shadow, and Action & Alignment.
