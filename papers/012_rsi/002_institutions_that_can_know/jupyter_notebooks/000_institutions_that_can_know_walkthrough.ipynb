# Institutions That Can Know

Exact finite examples for version 0.1.0. A recovery result concerns declared questions; it does not establish truth or permission. Run from the package root or this notebook directory.

```python
from pathlib import Path
import sys, itertools
root = Path.cwd()
if not (root / 'code/institutional_reasoning.py').is_file():
    root = root.parent
assert (root / 'code/institutional_reasoning.py').is_file()
sys.path.insert(0, str(root / 'code'))
from institutional_reasoning import recovery_decoder, side_information_code, jointly_acceptable
records = tuple(itertools.product((0, 1), repeat=3))
assert recovery_decoder(records, sum, sum) is not None
assert recovery_decoder(records, sum, lambda x: x) is None
print('Sum survives compression; order does not.')
```

## Restore the declared distinction

Each compressed fiber needs different codes only for its different answer profiles. The shared decoding table is excluded from the bit cost.

```python
bits, encoder, decoder = side_information_code(records, sum, lambda x: x)
assert bits == 2
assert all(decoder[sum(x), encoder[x]] == x for x in records)
print({'fixed_side_bits': bits, 'source_records': len(records)})
```

## Local consistency and global obstruction

```python
constraints = (lambda a,b,c: a == b, lambda a,b,c: b == c, lambda a,b,c: c != a)
for i,j in itertools.combinations(range(3),2):
    assert any(constraints[i](*x) and constraints[j](*x) for x in records)
assert not any(all(f(*x) for f in constraints) for x in records)
print('Every constraint pair is satisfiable; all three together are not.')
```

## Shared action leaves explanations separate

An intersection says which action meets the supplied constraints. An actual executor still needs a separately verified mandate.

```python
engineering = {'temporary_shutdown', 'mechanical_inspection'}
operations = {'temporary_shutdown', 'inventory_check'}
common = jointly_acceptable([engineering, operations])
assert common == {'temporary_shutdown'}
print(sorted(common))
```
