# Exact Bell certificates
Run from this notebook directory or the package root. No floating solver is trusted by the checker.

```python
from pathlib import Path
import sys
root = Path.cwd() if (Path.cwd()/"code").is_dir() else Path.cwd().parent
sys.path.insert(0,str(root/"code"))
from fractions import Fraction as Q
from bell_certificates import check,isotropic
check(*isotropic(Q(3,4)))
```

```python
result = check(*isotropic(Q(27,50)), uncertainty=Q(1,50))
assert result["distance"] == Q(1,50)
assert result["guaranteed_distance"] == 0
result
```

```python
p,q,c,r = isotropic(Q(3,4))
try:
    check(p,q,[2*x for x in c],r)
except ValueError as error:
    print("Rejected invalid witness:",error)
else:
    raise AssertionError("invalid witness accepted")
```
