### 1. Question and Prediction

> **Question:** When an equivalence relation groups individual records into classes, when can an evaluator be evaluated on the quotient classes?

**Prediction:** If an equivalence class contains both passing ($1$) and failing ($0$) records, can the evaluator assign a well-defined binary verdict to that class?

The answer is **no**: an evaluator descends to the quotient if and only if it is **constant on every equivalence class**. A mixed class constitutes a quotient obstruction.

```python
from io import BytesIO
from IPython.display import Image, display
import matplotlib.pyplot as plt

def display_figure(fig):
    try:
        with BytesIO() as buffer:
            fig.savefig(buffer, format='png', dpi=160, bbox_inches='tight')
            display(Image(data=buffer.getvalue()))
    finally:
        plt.close(fig)

# Concrete Records with Individual Verdicts
records = {
    'r1': 1,
    'r2': 0,
    'r3': 1,
    'r4': 1,
}

# Partition A: Groups r1 and r2 together into Class 1; r3 and r4 into Class 2
partition_coarse = {
    'Class 1': ['r1', 'r2'],
    'Class 2': ['r3', 'r4'],
}

# Audit quotient descent
def audit_quotient(records_dict, partition):
    descends = True
    issues = []
    for cname, members in partition.items():
        verdicts = {records_dict[m] for m in members}
        if len(verdicts) > 1:
            descends = False
            issues.append(f'{cname} is mixed with verdicts {verdicts}')
    return descends, issues

ok, issues = audit_quotient(records, partition_coarse)
print('Coarse Partition Audit:')
print(f'  Descends to quotient: {ok}')
print(f'  Issues:               {issues}')
```

### 3. Partition Refinement vs. Evaluator Modification

When faced with a mixed class, there are two fundamentally different interventions:

1. **Partition Refinement:** Subdivide the equivalence relation so that $r_1$ and $r_2$ belong to separate classes. This preserves the underlying empirical records.
2. **Evaluator Modification:** Changing the evaluator to artificially force agreement. This modifies the specification, which cannot retroactively turn historical failure evidence into success.

```python
# Refined partition separates r1 and r2
partition_refined = {
    'Class 1A': ['r1'],
    'Class 1B': ['r2'],
    'Class 2':  ['r3', 'r4'],
}

ok_ref, issues_ref = audit_quotient(records, partition_refined)
print('Refined Partition Audit:')
print(f'  Descends to quotient: {ok_ref}')
print(f'  Issues:               {issues_ref}')
for cname, members in partition_refined.items():
    v = records[members[0]]
    print(f'  {cname}: Members {members} -> Well-defined quotient verdict = {v}')
```

### 4. Exercise, Worked Answer, and Evidence Boundary

**Exercise:**  
If an analyst alters the evaluator so that $r_2$ is assigned verdict $1$ instead of $0$, has the quotient obstruction been resolved?

---

**Worked Answer:**  
No. Altering the evaluator changes the question being asked; it creates a new synthetic evaluator. The original evaluator remains obstructed on the coarse partition. Falsification records cannot be erased by changing the test retroactively.

---

**Evidence Boundary:**  
This lesson formalizes the universal quotient obstruction theorem (Claims LGGR3-C08–C09). An evaluator descends to an equivalence quotient if and only if it is constant on fibers.
