# Local Green, Global Red IV: Finite-Stage Solvability and Completion-Relative Obstruction

Interactive companion notebook illustrating:
1. **Finite-Stage Solvability (Local Green):** The congruence $3x + 1 \equiv 0 \pmod{2^n}$ solves exactly at every finite stage $n \ge 1$.
2. **Stage Compatibility:** The partial sums $x_n = -\sum_{k=0}^{n-1} (-2)^k$ satisfy $x_{n+1} \equiv x_n \pmod{2^n}$, defining a coherent thread in $\varprojlim \mathbb{Z}/2^n\mathbb{Z}$.
3. **Original-Carrier Failure (Global Red):** No integer in $\mathbb{Z}$ satisfies all finite stages simultaneously.
4. **Completion Solvability (Completion Green):** The thread converges $2$-adically to the unique root $x_\infty = -1/3 \in \mathbb{Z}_2$.
5. **Reflection Descent:** Completion green does not promote to original-carrier green.

```python
from pathlib import Path
import sys
from fractions import Fraction

candidates = [Path.cwd(), *Path.cwd().parents]
package_root = next(p for p in candidates if (p / "code" / "lggr_iv_core.py").is_file())
sys.path.insert(0, str(package_root / "code"))

from lggr_iv_core import (
    WitnessClassification,
    compute_x_n,
    verify_finite_stage,
    verify_compatibility,
    get_residue_thread,
    find_failing_stage_for_integer,
    adic_distance_to_completion_solution,
    classify_rational_witness,
)

print(f"Loaded LGGR IV core apparatus from: {package_root}")
```

## 1. Finite-Stage Solvability (Local Green)

We evaluate the exact identity $3x_n + 1 = (-2)^n$ for the first 8 stages:

```python
print(f"{'Stage n':<8} {'x_n':<12} {'3*x_n + 1':<14} {'(-2)^n':<12} {'Modulus':<10} {'Residue':<10} {'Green?'}")
print("-" * 76)
for n in range(1, 9):
    rec = verify_finite_stage(n)
    print(f"{rec.stage:<8} {rec.x_n:<12} {rec.lhs:<14} {rec.rhs:<12} {rec.modulus:<10} {rec.residue:<10} {rec.is_locally_green}")
```

## 2. Transition Compatibility Across Stages

We verify that $x_{n+1} \equiv x_n \pmod{2^n}$ for successive stages:

```python
for n in range(1, 9):
    diff, mod, is_comp = verify_compatibility(n)
    print(f"Stage {n} -> {n+1}: x_{{{n+1}}} - x_{{{n}}} = {diff:<8} mod {mod:<6} == 0? {is_comp}")
```

## 3. Original-Carrier Failure (Global Red)

No ordinary integer in $\mathbb{Z}$ satisfies all stages. For any tested integer, we find the first finite stage where the congruence fails:

```python
sample_ints = [0, 1, -1, 5, -3, 21, 100, -100]
for z in sample_ints:
    fail_stage = find_failing_stage_for_integer(z, search_limit=20)
    mod = 2**fail_stage
    val = (3 * z + 1) % mod
    print(f"Integer z = {z:<5} fails at stage n = {fail_stage:<2} (3*{z}+1 = {3*z+1} != 0 mod {mod}, remainder = {val})")
```

## 4. Completion Solvability and Reflection Descent

In $\mathbb{Z}_2$, $3$ is a unit, yielding the unique root $x_\infty = -1/3$.
We check the $2$-adic metric distance $|x_n - (-1/3)|_2 = 2^{-n}$ and classify candidates under reflection descent:

```python
print("2-adic Convergence:")
for n in range(1, 9):
    d = adic_distance_to_completion_solution(n)
    print(f"Stage {n}: |x_{n} - (-1/3)|_2 = {d} = {float(d):.6f}")

print("\nReflection Descent Classification:")
sol_completion = Fraction(-1, 3)
sol_class = classify_rational_witness(sol_completion)
print(f"Candidate -1/3 classification: {sol_class.value}")
print(f"Does completion solution descend to Z? {sol_class == WitnessClassification.ORIGINAL_CARRIER}")
```
